### The Brief

You are a data practitioner working with a community outpatient clinic. The clinic director has asked you to analyse two years of visit data. There are two decisions riding on your work:

1. **Staffing:** The director is considering hiring a care coordinator. The board needs data to justify that decision.
2. **Funding:** A county funder wants evidence of how the clinic serves uninsured patients. The analysis must use a consistent, defensible definition of uninsured and show county-level breakdown.

The clinic's questions from the ASK step were:
- Who are our repeat visitors — patients returning within 30 days?
- How are uninsured patients using the clinic?
- Which counties have the highest need?

**Your dataset:** 797 clean rows. Columns available:

| Column | What it contains |
|--------|------------------|
| `patient_id` | Unique patient identifier |
| `visit_id` | Unique visit identifier |
| `visit_date` | Date of visit |
| `visit_type` | Initial Visit, Follow-Up, Telehealth, Urgent Care |
| `insurance_type` | Medicare, Medicaid, Private, Uninsured, Tricare, Self-Pay |
| `icd_code` / `icd_description` | Diagnosis code and description |
| `provider_id` | Treating provider |
| `county` | Patient county |
| `gender` | Patient gender |
| `age` | Patient age (NaN where out of valid range) |
| `copay_amount` | Copay charged (NaN where invalid) |
| `follow_up_required` | Whether follow-up was flagged |
| `county_known` | Boolean — True if county is reliably recorded |
